```python
import tkinter as tk

# Create the main window
window = tk.Tk()
window.title("Calculator")
window.geometry("320x350")

# Entry widget to display expressions
entry = tk.Entry(window, width=16, font=('Arial', 24), borderwidth=6, relief="solid", justify='right',bg='lightblue', fg='black')
entry.grid(row=0, column=0, columnspan=4, padx=10, pady=10)

# Function to update the entry
def click(symbol):
    current = entry.get()
    entry.delete(0, tk.END)
    entry.insert(0, current + str(symbol))

# Function to evaluate the expression
def calculate():
    try:
        result = eval(entry.get())
        entry.delete(0, tk.END)
        entry.insert(0, str(result))
    except Exception as e:
        entry.delete(0, tk.END)
        entry.insert(0, "Error")

# Function to clear the entry
def clear():
    entry.delete(0, tk.END)

# Button layout
buttons = [
    ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
    ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
    ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
    ('0', 4, 0), ('.', 4, 1), ('+', 4, 2), ('=', 4, 3),
    ('Clear', 5, 0)
]

# Create and place buttons on the window
for (text, row, col) in buttons:
    if text == '=':
        btn = tk.Button(window, text=text, width=5, height=2, command=calculate)
    elif text == 'Clear':
        btn = tk.Button(window, text=text, width=40, height=2, command=clear)
        btn.grid(row=row, column=col, columnspan=40, padx=5, pady=5)
        continue
    else:
        btn = tk.Button(window, text=text, width=5, height=2, command=lambda t=text: click(t))
    btn.grid(row=row, column=col, padx=5, pady=5)

# Start the GUI event loop
window.mainloop()
```
